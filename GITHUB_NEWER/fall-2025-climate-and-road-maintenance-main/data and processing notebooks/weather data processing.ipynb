```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load data
weather = pd.read_csv("weather_hourly_1.csv")

# Ensure datetime and numeric columns
weather['time'] = pd.to_datetime(weather['time'], errors='coerce')
weather['snowfall (cm)'] = pd.to_numeric(weather['snowfall (cm)'], errors='coerce')

# Define half-month bins from Jan 2016 to end of 2025
bins = pd.date_range("2016-01-01", "2026-01-01", freq="15D")

# Aggregate snowfall by bin
snowfall_sum, _ = np.histogram(weather['time'], bins=bins, weights=weather['snowfall (cm)'])
snowfall_count, _ = np.histogram(weather['time'], bins=bins)

# Optional: compute average snowfall per hour if needed
# snowfall_avg = np.divide(snowfall_sum, snowfall_count, out=np.zeros_like(snowfall_sum), where=snowfall_count>0)

# Plot total snowfall per half-month bin
plt.figure(figsize=(12, 5))
plt.bar(
    bins[:-1],
    snowfall_sum,
    width=np.diff(bins).astype("timedelta64[D]").astype(int),
    align="edge",
    color="skyblue",
)
plt.title("Total Snowfall (cm) per Half-Month — 2016–2025")
plt.xlabel("Date")
plt.ylabel("Snowfall (cm)")
plt.xticks(pd.date_range("2016-01-01", "2026-01-01", freq="6M"), rotation=45)
plt.tight_layout()
plt.show()
```
