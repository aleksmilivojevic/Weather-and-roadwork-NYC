# Archived decision-tree experiment

This is an older decision-tree experiment kept for reference. It is not part of the main workflow.


```python
import pandas as pd
```

```python
df = pd.read_csv('data/derived/decisiontree_cleaned_data.csv')
```

```python
df
```

```python
"""
Not a time series analysis; (street,year) is each treated as independent data point
Want to regress [roadwork_factor] onto [snowfall, total_area, average_coverage]
OLS linear regression gives R^2 of 0.237

snowfall: amount of snowfall on that street in that year 
total_area: total area of the street
roadwork_factor: area of roadwork done on given street in given year, divided by total_area
average_coverage: normalized plow coverage computed from snaps and last_visited
"""
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
```

```python
model = DecisionTreeRegressor(max_depth = 6, random_state=0)
```

```python
X = df[['snowfall', 'total_area', 'average_coverage']]
y = df['roadwork_factor']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=0)
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error, r2_score

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
```

```python
print(mse)
print(r2)
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
randomforestmodel = RandomForestRegressor()
```

```python
randomforestmodel.fit(X_train, y_train)
randomforest_pred = randomforestmodel.predict(X_test)

print(mean_squared_error(randomforest_pred, y_test))
print(r2_score(randomforest_pred, y_test))
```
